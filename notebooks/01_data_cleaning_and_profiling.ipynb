### 1. Ingestion & Library Setup

```python
import os
import pandas as pd
import numpy as np

data_dir = '../data/processed'
dim_content = pd.read_csv(os.path.join(data_dir, 'dim_content.csv'))
dim_users = pd.read_csv(os.path.join(data_dir, 'dim_users.csv'))
dim_date = pd.read_csv(os.path.join(data_dir, 'dim_date.csv'))
fact_streams = pd.read_csv(os.path.join(data_dir, 'fact_streams.csv'))

print(f'Content: {dim_content.shape}')
print(f'Users: {dim_users.shape}')
print(f'Date: {dim_date.shape}')
print(f'Streams: {fact_streams.shape}')
```

### 2. Missing Value & Cardinality Profiling

```python
print('--- Missing Values in Streams ---')
print(fact_streams.isnull().sum())

print('\n--- Unique User Regions ---')
print(dim_users['region'].value_counts())

print('\n--- Content by Primary Genre ---')
print(dim_content['primary_genre'].value_counts())
```

### 3. Data Integrity & Type Validation

```python
fact_streams['stream_date'] = pd.to_datetime(fact_streams['stream_date'])
dim_users['signup_date'] = pd.to_datetime(dim_users['signup_date'])

assert (fact_streams['completion_percentage'] <= 100).all(), 'Completion rate exceeds 100%!'
assert (fact_streams['completion_percentage'] >= 0).all(), 'Negative completion rate!'
print('Data integrity assertions passed successfully!')
```
