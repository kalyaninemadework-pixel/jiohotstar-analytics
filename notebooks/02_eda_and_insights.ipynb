### 1. Genre Efficiency Analysis

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data_dir = '../data/processed'
dim_content = pd.read_csv(os.path.join(data_dir, 'dim_content.csv'))
fact_streams = pd.read_csv(os.path.join(data_dir, 'fact_streams.csv'))

merged = fact_streams.merge(dim_content, on='content_id')
genre_perf = merged.groupby('primary_genre').agg(
    total_streams=('stream_id', 'count'),
    total_hours=('watch_duration_minutes', lambda x: x.sum()/60.0),
    avg_comp=('completion_percentage', 'mean')
).reset_index()

genre_perf.sort_values(by='total_hours', ascending=False)
```

### 2. Language Consumption & South Indian Regional Stickiness

```python
lang_perf = merged.groupby('original_language').agg(
    titles=('content_id', 'nunique'),
    total_hours=('watch_duration_minutes', lambda x: x.sum()/60.0),
    avg_comp=('completion_percentage', 'mean')
).reset_index().sort_values(by='total_hours', ascending=False)

print(lang_perf)
```
